## Initial Findings

Based on the initial dataset inspection:

- The dataset contains **148,670 rows and 34 columns**.
- The target variable is **Status**, where:
  - **0:** Loan not defaulted
  - **1:** Loan defaulted
- The dataset contains **13 numerical columns and 21 categorical columns** based on their current data types.
- Several features contain missing values, with the highest number of missing values found in:
  - `Upfront_charges`
  - `Interest_rate_spread`
  - `rate_of_interest`
  - `dtir1`
  - `property_value`
  - `LTV`
  - `income`
- The dataset contains **0 duplicate rows**.
- The target distribution is moderately imbalanced:
  - **Class 0:** 112,031 observations (75.36%)
  - **Class 1:** 36,639 observations (24.64%)
- Several categorical features will require encoding before machine learning model training.
- The `year` column contains only one unique value (2019), so it provides no variation and may be considered for removal during feature selection.
- The `ID` column appears to be an identifier rather than a predictive feature and should be investigated before modeling.
- Further EDA is required to investigate numerical distributions, categorical feature distributions, outliers, multicollinearity, feature relationships with the target, and potential data leakage.
